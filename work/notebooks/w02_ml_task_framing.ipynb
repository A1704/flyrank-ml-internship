## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

For the starter dataset, I will create is_declining_label from the existing trend_direction field as a proxy target. It represents whether a content item is currently labeled as declining, rather than a future outcome. Because this proxy is derived from trend_direction and ultimately trend_pct, those fields should not be used as model features. For a stronger future-looking version of this task, I would prefer a target based on an observed outcome in a later time window.

```python
import pandas as pd

df = pd.read_csv("flyrank-ml-internship/data/raw/content_refresh_anonymized.csv")

# Create the starter proxy target from the existing trend_direction field
df["is_declining_label"] = df["trend_direction"].str.lower().eq("down").astype(int)

print("Rows:", len(df))
print("Declining proxy rate:", round(df["is_declining_label"].mean(), 3))
```
